```python
from typing import TypedDict


class State(TypedDict):
    message: str
    response: str

# this is state Think of State as the information carried through the graph.
```

```python
def greet_node(state: State):
    print("Running greet_node...")

    state["response"] = f"Hello {state['name']}!"

    return state

# node 1 greet_node
```

```python
def hospital_node(state: State):
    print("Running hospital_node...")

    state["response"] += " Welcome to the Hospital AI Assistant."

    return state

# node 2 hospital_node
```

```python
# graph 

from langgraph.graph import StateGraph


builder = StateGraph(State)
```

```python
# in the graph we adding node 

builder.add_node("greet", greet_node)
builder.add_node("hospital", hospital_node)
```

```python
# adding edge between nodes 
from langgraph.graph import END, START



builder.add_edge(START, "greet")
builder.add_edge("greet", "hospital")
builder.add_edge("hospital", END)
```

```python
graph = builder.compile()
```

```python
result = graph.invoke({
    "name": "Thangarasu",
    "response": ""
})

result
```
